# F strings
FStrings are the preferred way to intrerpolate strings with values because of its high readability

```python
val = "Hello, World!"
print(f"Some said: \"{val}\" because it is common knowledge that 3+4 = {3+4}")
print("Some said: \"" + val + "\" because it is common knowledge that 3+4 = " + str(3+4))
```

# Type Hints 
Type hints are not used at run-time (so they are not strict) but improve human readability and let ides provide:
- static analysis (to anticipate errors)
- suggestions for developer
- docs

```python
# without type hints and fstrings
def say_hi(name):
    return "Hi " + name

# with type hints and fstrings
def say_hi_better(name: str) -> str:
    return f"Hi {name}"

print(say_hi("Alice"))
print(say_hi_better("Bob"))
```

Type hints can also suggest if a field is optional or should only have specific values (these are not enforced at run time by python)

```python
from typing import Optional, Literal

def say_hi_v2(
        name_optional: str = "GuessWho",
        name_optional_2: str | None = None,
        name_optional_3: Optional[str] = None,
        only_some: Literal["some", "only"] = "some"):
    print (f"name_optional: {name_optional},\r\nname_optional_2: {name_optional_2},\r\nname_optional_3: {name_optional_3},\r\nonly_some: {only_some}\r\n")

print(say_hi_v2(name_optional_3="Charlie"))
```
