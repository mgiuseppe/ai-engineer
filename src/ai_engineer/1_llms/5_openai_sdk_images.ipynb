## Pass image as URL

```python
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()  # Load environment variables from .env file

client = OpenAI()

response = client.responses.create(
    model="gpt-5-nano",
    reasoning={"effort": "minimal"},
    store=False,
    input=[
        {
            "role": "user",
            "content": [ # Text, image, or audio input to the model, used to generate a response. Can also contain previous assistant responses.
                {"type": "input_text", "text": "Describe this image"},
                {"type": "input_image", "image_url": "https://picsum.photos/200"},
            ]
        }
    ],
)

print(f"output text: {response.output_text}")
print(f"usage - input tokens: {response.usage.input_tokens} - output tokens: {response.usage.output_tokens} - total: {response.usage.total_tokens}")
```

## Pass image as Base64

```python
import base64

# Function to encode the image
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")

# Getting the Base64 string
base64_image = encode_image(image_path = "200x200.jpg")

response = client.responses.create(
    model="gpt-5-nano",
    reasoning={"effort": "minimal"},
    store=False,
    input=[
        {
            "role": "user",
            "content": [
                {"type": "input_text", "text": "Describe this image"},
                {"type": "input_image", "image_url": f"data:image/jpeg;base64,{base64_image}", "detail": "auto"},
            ]
        }
    ],
)

print(f"output text: {response.output_text}")
print(f"usage - input tokens: {response.usage.input_tokens} - output tokens: {response.usage.output_tokens} - total: {response.usage.total_tokens}")
```

## Generate an Image

```python
from openai import OpenAI
import base64

client = OpenAI()

response = client.responses.create(
    model="gpt-5-nano",
    input="Generate an image of a person standing in a wide, open desert or beach landscape. They are facing away from the camera with arms slightly outstretched, wearing a light-colored dress. The background shows a clear sky and low dunes or rocks on the horizon, giving a sense of spaciousness and a sunny day.",
    tools=[{"type": "image_generation", "size": "1024x1024"}]
)

print(response.usage)

# Save the image to a file
image_data = [output.result for output in response.output if output.type == "image_generation_call"]
if image_data:
    image_base64 = image_data[0]
    with open("out_girl.png", "wb") as f:
        f.write(base64.b64decode(image_base64))
```
