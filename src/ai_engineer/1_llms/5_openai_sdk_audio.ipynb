# OpenAI SDK - Multi modal LLMs - Audio

- source: [Audio Guide](https://developers.openai.com/api/docs/guides/audio)

OpenAI covers a lot of use cases supporting WebRTC, WebSocket and SIP (phone calls):
- Speech Generation (TTS models)
- Voice Agents (Speech-To-Speech with reason, talk, tool calling with realtime models)
- Live translation (With both audio and subtitles with realtime models)
- Transcription (With realtime or filetranscription)

According to your needs you can use a different model or api (realtime, audio, chat.completions), see the doc for further info.

Here we will see just TTS and Audio Responses.

### Client.Audio - Generate mp3 TTS (gpt-4o-mini-tts) 

TTS model. Generates a voice telling the exact content passed in input.

```python
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()
client = OpenAI()

speech_file_path = "mentor.mp3"
with client.audio.speech.with_streaming_response.create(
    model="gpt-4o-mini-tts",
    voice="onyx",
    input="Hello my friends, I believe your training is complete. You're not yet true heroes, untested by battle as you are. But I sense in you great potential. Let me tell you of the foe you face, Zargon.",
    instructions="Speak in a slow and powerful tone.",
) as response:
    response.stream_to_file(speech_file_path)
    # no usage with client.audio
```

### Client.Chat.Completions - Generate Audio Answer (gpt-audio-1.5)

Not a TTS Model. Will generate an audio answer to the input.

Note that we are using the chat completions API because this is not available in Responses API. 

```python
import base64
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()
client = OpenAI()

completion = client.chat.completions.create(
    model="gpt-audio-1.5",
    modalities=["text", "audio"],
    audio={"voice": "onyx", "format": "wav"},
    messages=[
        {
            "role": "user", 
            "content": "Hello, how are you doing?"
        }
    ],
)
wav_bytes = base64.b64decode(completion.choices[0].message.audio.data)
with open("mentor.wav", "wb") as f:
    f.write(wav_bytes)
```
