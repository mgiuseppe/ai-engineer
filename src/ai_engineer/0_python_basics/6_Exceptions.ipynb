# Exceptions

Exceptions are handled in try / except blocks:
- try: code that might raise an error goes here.
- except: runs if a matching exception is raised.
- else: runs only if the try block succeeded (no exception).
- finally: always runs — typically used for cleanup whether or not an error occurred.

```python
class MyCustomError(Exception):
    """Raised when I think it's okay to raise it."""
    pass

try:
    raise MyCustomError("I need to raise this now")
except MyCustomError as e:
    print(f"Error catched: {e}")
else:
    print("Runs only if the try block succeeded (no exception)")
finally:
    print("Always runs")
```

```python
try: 
    try:
        f = open("data.txt")
    except MyCustomError:
        print("Error catched.")
    else:
        print(f.read())
        f.close()
except FileNotFoundError:
    print("File not found.")
```
