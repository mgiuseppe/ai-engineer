# Context Managers

Useful to ensure resources will be closed once done, just like the .net using construct

```python
import httpx
with httpx.Client() as client:
    r = client.get("https://example.com")
    print(f'response = {r}')
# client.close() called automatically here, even if an exception occurred

with open(".env") as f:
    data = f.read()
    print(f'data = {data}')
# file automatically closed here, even if read() raises
```
