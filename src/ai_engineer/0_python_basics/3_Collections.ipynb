# Collections

## Lists

```python
fruits = ["Apple", "Banana", "Melon", "Cherry"]
print(fruits)

fruits.append("Orange")
fruits.remove("Apple")
print(fruits)

fruits.sort()
print(fruits)
```

## Tuples

```python
rgb = (255, 255, 10)
print(rgb[0])
print(rgb[2])
print(rgb)
```

## Sets

```python
set_example = {1, 2, 3, 4, 5}
set_2 = {8}

set_example.add(6)
set_example.remove(3)
set_example.discard(10)  # Does not raise an error if the element is not found
print(set_example | set_2)  # Returns a new set with elements from both sets

set_3 = set_example.union(set_2)  # Returns a new set with elements from both sets
print(set_example)
print(set_2)
print(set_3)
```

## Dictionaries

```python
lookup = {"name": "John", "age": 30, "city": "New York"}
print(lookup["name"])  # Accessing value by key
print(lookup["age"])   # Accessing value by key
print(lookup["city"])  # Accessing value by key
```

## Numpy Arrays
