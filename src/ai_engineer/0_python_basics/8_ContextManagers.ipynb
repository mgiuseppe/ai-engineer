# Context Managers

Useful for ensuring resources are closed once done, just like the .NET "using" construct

With statement

```python
import httpx

with httpx.Client() as client:
    r = client.get("https://example.com")
# client.close() called automatically here, even if an exception occurred

with open(".env") as f:
    data = f.read()
# file automatically closed here, even if read() raises
```

Async with statement

```python
async with httpx.AsyncClient() as client:
    r = await client.get("https://example.com")
# client.close() called automatically here, even if an exception occurred
```

Build your own context manager supporting the "with" construct

```python
from contextlib import contextmanager
import time

@contextmanager
def timer(label):
    start = time.perf_counter()
    yield # the code inside the with block will be executed here
    print(f"{label}: {time.perf_counter() - start:.2f}s")

with timer("my block"):
    time.sleep(2)
```
