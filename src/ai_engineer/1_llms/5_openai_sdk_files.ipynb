# OpenAI SDK - Multi modal LLMs - Files

- source: [Files inputs](https://developers.openai.com/api/docs/guides/file-inputs)  
- list of supported files and mime types: [list](https://developers.openai.com/api/docs/guides/file-inputs?api-mode=responses#full-list-of-accepted-file-types)

Supported files as of today (02/09/2026):
- PDF files	.pdf
- Text and code	.txt, .md, .json, .html, .xml, code files
- Rich documents	.doc, .docx, .rtf, .odt
- Presentations	.ppt, .pptx
- Spreadsheets	.csv, .xls, .xlsx

## Pass file as URL

```python
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()  # Load environment variables from .env file

client = OpenAI()

response = client.responses.create(
    model="gpt-5-nano",
    input=[
        {
            "role": "user",
            "content": [
                {
                    "type": "input_text",
                    "text": "Analyze the letter and provide a summary of the key points.",
                },
                {
                    "type": "input_file",
                    "file_url": "https://www.berkshirehathaway.com/letters/2024ltr.pdf",
                },
            ],
        },
    ],
)

print(f"usage: input: {response.usage.input_tokens} - output: {response.usage.output_tokens} - total: {response.usage.total_tokens}")
print(f"output text: {response.output_text}")
```

### Pass file with Files API

```python
file = client.files.create(file=open("sales_report.xlsx", "rb"), purpose="user_data")

response = client.responses.create(
    model="gpt-5.6",
    input=[
        {
            "role": "user",
            "content": [
                {
                    "type": "input_file",
                    "file_id": file.id,
                },
                {
                    "type": "input_text",
                    "text": "Who is the best salesperson?",
                },
            ],
        }
    ],
)

print(f"usage: input: {response.usage.input_tokens} - output: {response.usage.output_tokens} - total: {response.usage.total_tokens}")
print(response.output_text)
```

### Pass file as Base64

```python
import base64

with open("anonymous_title.pptx", "rb") as f:
    data = f.read()
base64_string = base64.b64encode(data).decode("utf-8")

response = client.responses.create(
    model="gpt-5-nano",
    input=[
        {
            "role": "user",
            "content": [
                {
                    "type": "input_file",
                    "filename": "anonymous_title.pptx",
                    "file_data": f"data:application/vnd.openxmlformats-officedocument.presentationml.presentation;base64,{base64_string}",
                },
                {
                    "type": "input_text",
                    "text": "What is the file about? Is it complete? Give just a short answer, no more than 100 words.",
                },
            ],
        },
    ],
)

print(f"usage: input: {response.usage.input_tokens} - output: {response.usage.output_tokens} - total: {response.usage.total_tokens}")
print(response.output_text)
```
