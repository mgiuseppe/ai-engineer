# Iterable and Iterator

An iterable is a container you can loop over; an iterator is the thing actually doing the looping, one step at a time.

* Iterable = "I can be iterated" (List, string, file, ...)
* Iterator = "I am the object being iterated" (has a cursor, moves forward, exhausted at the end)

## Iterable
* Anything with \_\_iter()\_\_, which returns an iterator.
* Represents a collection.
* Can be looped multiple times, each for loop asks for a brand new iterator.

## Iterator
* Anything with \_\_next\_\_() (and also \_\_iter\_\_() which returns itself) is an iterator.
* Represents progress through a collection, it holds the state (where it currently is).
* Gets exhausted: once you reach the end it's done and cannot be rewinded, so it can be looped only once.

```python
# iterable
class Countdown:
    def __init__(self, start):
        self.start = start
    def __iter__(self):
        return CountdownIterator(self.start)

# iterator (note that each iterator is also an iterable because it implements __iter__)
class CountdownIterator: 
    def __init__(self, current):
        self.current = current
    def __iter__(self):
        return self          # iterators return themselves
    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        self.current -= 1
        return self.current + 1

print("Countdown from 5:")
for n in Countdown(5):
    print(n)   # 5 4 3 2 1

print("From 1 to 3:")
numbers = [1, 2, 3] # iterable
for n in numbers:  # for loop asks for the iterator calling iter(number) and loops through it
    print(n)  # 1 2 3
```

## Generator Function
* Shortcut to define an iterable and an iterator by just defining a function
* The function contains yield keyword

```python
def countdown(start):
    current = start
    while current > 0:
        yield current
        current -= 1

for n in countdown(5):
    print(n)   # 5 4 3 2 1
```
