## LIST COMPREHENSIONS
A list comprehension is a concise way to create lists. It consists of square brackets containing: 
- an expression 
- followed by a for clause
- then zero or more for or if clauses

The expressions can be anything, meaning you can put in all kinds of objects in lists.

```python
#Example 1: Create a list of squares of numbers from 0 to 9
squares = [x**2 for x in range(10)]
print(f"squares using list comprehension: {squares}")  # Output: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]

#Example 2: Create a list of even numbers from 0 to 9
even_numbers = [x for x in range(10) if x % 2 == 0]
print(f"even numbers using list comprehension: {even_numbers}")  # Output: [0, 2, 4, 6, 8]

#Example 3: Create a list of tuples (number, square) for numbers from 0 to 9
number_square_tuples = [(x, x**2) for x in range(10)]
print(f"number-square tuples using list comprehension: {number_square_tuples}")  # Output: [(0, 0), (1, 1), (2, 4), (3, 9), (4, 16), (5, 25), (6, 36), (7, 49), (8, 64), (9, 81)]   
```


## GENERATOR EXPRESSIONS
- A generator expressions is similar to a list comprehension but uses parentheses instead of square brackets.
- It produces a generator object (iterator) that produces items on demand, instead of creating a list at once.
- This can be more memory efficient for large datasets but can be consumed only once and can't be random accessed or counted.


```python
#Example 4: Using a generator expression to create a generator of squares of numbers from 0 to 9
squares_generator = (x**2 for x in range(10))
print(f"squares using generator: {squares_generator}")  # Output: <generator object <genexpr> at 0x...>
print(f"squares using generator: {list(squares_generator)}")  # Output: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
print(f"squares using generator: {list(squares_generator)}")  # Output: [] (the generator is exhausted after the first iteration)
```


## MAP AND FILTER
The map() function applies a given function to each item of an iterator and returns a map object (an iterator).

The filter() function constructs an iterator from elements of an iterator for which a function returns true.

```python
#Example 5: Using map to create a list of squares of numbers from 0 to 9
squares_map = map(lambda x: x**2, range(10))
print(f"squares using map: {squares_map}") # Output: <map object at 0x...>
print(f"squares using map: {list(squares_map)}")  # Output: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
print(f"squares using map: {list(squares_map)}")  # Output: [] (the map object is exhausted after the first iteration)

#Example 6: Using filter to create a list of even numbers from 0 to 9
even_numbers_filter = filter(lambda x: x % 2 == 0, range(10))
print("")
print(f"even numbers using filter: {even_numbers_filter}")  # Output: <filter object at 0x...>
print(f"even numbers using filter: {list(even_numbers_filter)}")  # Output: [0, 2, 4, 6, 8]
print(f"even numbers using filter: {list(even_numbers_filter)}")  # Output: [] (the filter object is exhausted after the first iteration)
```

## ZIP AND UNZIP
The zip() function takes iterables (can be zero or more), aggregates them in a tuple, and returns an iterator.

```python
#Example 7: Using zip to combine two lists into a list of tuples
names = ["Alice", "Bob", "Charlie", "David"]  # Note: names has one more element than ages
ages = [25, 30, 35]
zipped = zip(names, ages)
print(f"zipped names and ages: {zipped}")  # Output: <zip object at 0x...>
print(f"zipped names and ages: {list(zipped)}")  # Output: [('Alice', 25), ('Bob', 30), ('Charlie', 35)]
print(f"zipped names and ages: {list(zipped)}")  # Output: [] (the zip object is exhausted after the first iteration)

#Example 8: Using zip with the unpacking operator (*) to unzip a list of tuples into two separate lists
zipped_list = [('Alice', 25), ('Bob', 30), ('Charlie', 35)]
unzipped_names, unzipped_ages = zip(*zipped_list) # zip(*) produces an iterator but assigning it to two variables automatically consumes the iterator
print("")
print(f"unzipped names: {unzipped_names}")  # Output: ('Alice', 'Bob', 'Charlie')
print(f"unzipped ages: {unzipped_ages}")  # Output: (25, 30, 35)
```


## ENUMERATE
The enumerate() function adds a counter to an iterable and returns it as an enumerate object (an iterator of tuples containing the index and the value).

```python
#Example 9: Using enumerate to get the index and value of items in a list
fruits = ["apple", "banana", "cherry"]
enumerated_fruits = enumerate(fruits)
print("")
print(f"enumerated fruits: {enumerated_fruits}")  # Output: <enumerate object at 0x...>
print(f"enumerated fruits: {list(enumerated_fruits)}")  # Output: [(0, 'apple'), (1, 'banana'), (2, 'cherry')]
print(f"enumerated fruits: {list(enumerated_fruits)}")  # Output: [] (the enumerate object is exhausted after the first iteration)  

for i, nome in enumerate(fruits, start=1):
    print(f"enumerate in for: {i}, {nome}")  # Output: 1, apple; 2, banana; 3, cherry
```

## EXPERT TIPS
1. Use list comprehensions for simple transformations and filtering.
2. Use generator expressions when working with large datasets to save memory, as they return an iterator with lazy evaluation
    * Rember that you cannot iterate over an iterator more than once and you can't random access or count items 
3. Map() and Filter() are equivalent to generator expessions but less readable and pythonic.
4. Use zip() to combine multiple iterables into a single iterator of tuples, and use the unpacking operator (*) to unzip them back into separate iterables.
5. Use enumerate() to add index to values when iterating over an iterable, which can be more convenient than using a separate counter variable.
