# Classes

When you create a class you need to specify:
- __init__ (constructor)
- __repr__ (print - optional)
- __eq__ (comparison - optional)

```python
class Point:
    """This is a docstring for the Point class. It represents a point in 2D space with x and y coordinates."""
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

    def __eq__(self, other):
        return (self.x, self.y) == (other.x, other.y)

p1 = Point(10,20)
p2 = Point(10,20)
p3 = Point(60,70)

print(p1)
print(p2)
print(p3)
print(f"p1==p2: {p1==p2}")
print(f"p1==p3: {p1==p3}")
```

## Inheritance

```python
class ChildPoint(Point):
    """This is a docstring for the ChildPoint class. It inherits from Point and adds a z coordinate."""
    def __init__(self, x, y, z):
        super().__init__(x, y)
        self.z = z

    def __repr__(self):
        return f"ChildPoint(x={self.x}, y={self.y}, z={self.z})"

cp = ChildPoint(10,20,30)
print(cp)
```

# Data classes

You can also create a class using the @dataclass decorator. This will automatically create __init__ __repr__ and __eq__. 

Under the hood it's a common class (by default it's mutable).

This is commonly used when your class doesn't have complex logic but is only a bunch of attributes.

```python
from dataclasses import dataclass, field

@dataclass
class Point:
    x: int
    y: int
    z: int = field(default=50)
    tags: list = field(default_factory=list)  # mutable defaults need default_factory

p1 = Point (10, 20)
p2 = Point (10, 20, tags=["tag1", "tag2"])
p3 = Point (10, 20, tags=["tag1", "tag2"])

print(p1)
print(p2)
print(p3)
print(f"p1==p2: {p1==p2}")
print(f"p1==p3: {p1==p3}") # This will return False because the tags lists are different objects, even though they have the same content.
```
