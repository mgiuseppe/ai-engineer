# Pathlib

Pathlib is a module that introduces the Path class, letting you handle paths as objects (methods and / operator) in a cross-platform fashion

```python
from pathlib import Path

p = Path("../0_python_basics/1_FStrings_TypeHints.ipynb")

print(f"p.name: {p.name}")        # '1_FStrings_TypeHints.ipynb'
print(f"p.stem: {p.stem}")        # '1_FStrings_TypeHints'
print(f"p.suffix: {p.suffix}")    # '.ipynb'
print(f"p.parent: {p.parent}")    # Path('../0_python_basics')
print(f"p.parts: {p.parts}")      # ('..', '0_python_basics', '1_FStrings_TypeHints.ipynb')

print(f"p.exists(): {p.exists()}")
print(f"p.is_file(): {p.is_file()}")
print(f"p.is_dir(): {p.is_dir()}")

print("\r\nGlob:")
for f in Path("./").glob("*.ipynb"):    # iterate matching files
    print(f)
```
